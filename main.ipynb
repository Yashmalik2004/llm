```python
import pandas as pd

log_msgs = [
    "INFO: Server started successfully",
    "WARNING: Low disk space available",
    "ERROR: Database connection failed",
    "CRITICAL: Security breach detected",
    "INFO: New feature released",
]

df = pd.DataFrame(log_msgs, columns=["Message"], index = range(1, len(log_msgs)+1))
df
```

```python
df["Sentiment"] = df["Message"].apply(lambda x:llm.invoke(f"Classify the severity of the log messages, make give output as CRITICAL if the message if critical : {x}"))
df
```
